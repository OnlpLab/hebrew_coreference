```python
import os
gold_path = r"C:\Users\rafael\Desktop\studies\MSC\Theses\Bert NP result\input_sep"
res_path = r"C:\Users\rafael\Desktop\studies\MSC\Theses\Bert NP result\res"
```

```python
def read_file(path, sep="\t"):
    with open(path, encoding='utf-8') as f:
        lines = f.readlines()
    sentences = []
    sent = []
    for line in lines:
        if line == "" or line == "\n":
            if sent:
                sentences.append(sent)
                sent = []
        else:
            sent.append(tuple(line.strip().split(sep)))
    return sentences
```

```python
gold_test = read_file(os.path.join(gold_path, 'test.txt'), " ")
res_test = read_file(os.path.join(res_path,"1" ,'test_predictions.txt'), " ")
```

```python
to_print = []

for g, r in zip(gold_test, res_test):
    have_mistake = False
    for g_w, r_w in zip(g,r):
        word_g, tag_g = g_w
        word_r, tag_r = r_w
        # if word_g != word_r:
        #     raise Exception("hopa hopaaa")
        if tag_r != tag_g:
            have_mistake=True
            print(f"word {word_r}")
            print(f"tag gold: {tag_g} tag res {tag_r}")
            sent = " ".join([t[0] for t in g])

            to_print.append({"word": word_r, "gold": tag_g, "res": tag_r, "sent":  sent, "gold_tags": g, "res_tags": r})
    if have_mistake:
        print(sent)
        print("gold")
        print(g)
        print("result")
        print(r)
```

```python
import pandas as pd
pd.DataFrame(to_print).to_csv("analysis_np.csv", encoding='utf-8-sig')
```
