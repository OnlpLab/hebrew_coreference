## Third Round Agreement - Inner Agreement Point-AI

```python
conllu_annotated_files_path = "../../conllu_out_annotation/documents_262_to_290"
```

```python
file_by_annotator = defaultdict(list)
conllu_annotated_files = os.listdir(conllu_annotated_files_path)
for f in conllu_annotated_files:
    file_by_annotator[extract_name(f)].append(f)
keep_singletons = False
calc_with_mentions = False

Hadar_vs_Tal = two_annotators_agreement("Hadar", "Tal", keep_singletons, calc_with_mentions, file_by_annotator,
                                        conllu_annotated_files_path)

Itay_vs_Tal = two_annotators_agreement("Itay", "Tal", keep_singletons, calc_with_mentions, file_by_annotator,
                                       conllu_annotated_files_path)
Hadar_vs_Itay = two_annotators_agreement("Hadar", "Itay", keep_singletons, calc_with_mentions, file_by_annotator,
                                         conllu_annotated_files_path)

third_round_agreement = [
    ("Itay_vs_Tal No Singletons", Itay_vs_Tal),
    ("Hadar_vs_Tal No Singletons", Hadar_vs_Tal),
    ("Hadar_vs_Itay No Singletons", Hadar_vs_Itay),

]
```

```python
for name, score in third_round_agreement:
    print(name + " Conll score by muc, bcub, ceafe, lea  (without Mentions)")
    df = pd.DataFrame(score)
    df['mean'] = df.mean(axis=1)
    print(df['mean']['f1'])
```

```python
should_keep_singleton = False
include_mentions = False
scores = n_annotators_agreement(["Itay", "Hadar", "Tal"], should_keep_singleton, include_mentions,
                                file_by_annotator, conllu_annotated_files_path)
```

```python
print('Total score of "Itay", "Hadar", "Tal" with mentions: ')
data = {'values': [pd.DataFrame(score).loc['f1'].mean() for score in scores]}
df = pd.DataFrame(data)

print(df['values'].mean())
```

```python
should_keep_singleton = False
include_mentions = False
scores = n_annotators_agreement(["Hinoy", "Yechiel", "Hadas"], should_keep_singleton, include_mentions,
                                file_by_annotator, conllu_annotated_files_path)
```

```python
print('Total score of "Hinoy", "Yechiel", "Hadas" no mentions: ')
data_no_mentions = {'values': [pd.DataFrame(score).loc['f1'].mean() for score in scores]}
df_no_mentions = pd.DataFrame(data_no_mentions)
print(df_no_mentions['values'].mean())
```
