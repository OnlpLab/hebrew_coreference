## Third Round Agreement - Inner Agreement Point-AI

```python
conllu_annotated_files_path = "../conllu_out_annotation/first_batch"
file_by_annotator = defaultdict(list)
conllu_annotated_files = os.listdir(conllu_annotated_files_path)
for f in conllu_annotated_files:
    file_by_annotator[extract_name(f)].append(f)

keep_singletons = False
Arbel_vs_Ohad_wo_singleton = two_annotators_agreement("kovesh", "Leurer", False, True, file_by_annotator, conllu_annotated_files_path)
Arbel_vs_Ohad_w_singleton = two_annotators_agreement("kovesh", "Leurer", True,True, file_by_annotator, conllu_annotated_files_path)
Shaked_vs_Elisheva_wo_singleton = two_annotators_agreement("Greenfeld", "Jeffay", False, True, file_by_annotator, conllu_annotated_files_path)
Shaked_vs_Elisheva_w_singleton = two_annotators_agreement("Greenfeld", "Jeffay", True,True, file_by_annotator, conllu_annotated_files_path)
Arbel_vs_Elisheva_wo_singleton = two_annotators_agreement("kovesh", "Jeffay", False, True, file_by_annotator, conllu_annotated_files_path)
Arbel_vs_Elisheva_w_singleton = two_annotators_agreement("kovesh", "Jeffay", True,True, file_by_annotator, conllu_annotated_files_path)
Ohad_vs_Elisheva_wo_singleton = two_annotators_agreement("Leurer", "Jeffay", False, True, file_by_annotator, conllu_annotated_files_path)
Ohad_vs_Elisheva_w_singleton = two_annotators_agreement("Leurer", "Jeffay", True,True, file_by_annotator, conllu_annotated_files_path)
third_round_agreement = [
    ("Arbel vs. Ohad No Singletons", Arbel_vs_Ohad_wo_singleton),
    ("Arbel vs. Ohad  With Singletons", Arbel_vs_Ohad_w_singleton),
    ("Shaked vs. Elisheva No Singletons", Shaked_vs_Elisheva_wo_singleton),
    ("Shaked vs. Elisheva  With Singletons", Shaked_vs_Elisheva_w_singleton),

    ("Arbel vs. Elisheva  No Singletons",Arbel_vs_Elisheva_wo_singleton),
    ("Arbel vs. Elisheva  With Singletons",Arbel_vs_Elisheva_w_singleton),
    ("Ohad vs. Elisheva  No Singletons", Ohad_vs_Elisheva_wo_singleton),
    ("Ohad vs. Elisheva  With Singletons", Ohad_vs_Elisheva_w_singleton)]
```

```python
for name, score in third_round_agreement:
    print(name + " Conll score by muc, bcub, ceafe, lea  (without Mentions)")
    df = pd.DataFrame(score)
    df = df.drop(columns=["mentions"])
    df['mean'] = df.mean(axis=1)
    print(df['mean']['f1'])
for name, score in third_round_agreement:
    print(name + " Mention score")
    mentions = pd.DataFrame(score)["mentions"]
    print(mentions['f1'])
```
