```python
import os
from collections import defaultdict
import pandas as pd
from agreement_utils import extract_name, two_annotators_agreement, n_annotators_agreement
```

## Third Round Agreement - Inner Agreement Point-AI

```python
conllu_annotated_files_path = "../conllu_out_annotation/documents_35_to_47"
file_by_annotator = defaultdict(list)
conllu_annotated_files = os.listdir(conllu_annotated_files_path)
for f in conllu_annotated_files:
    file_by_annotator[extract_name(f)].append(f)
keep_singletons = False


Arbel_vs_Ido = two_annotators_agreement("Arbel", "Ido", keep_singletons,True, file_by_annotator, conllu_annotated_files_path, )
Arbel_vs_Elisheva = two_annotators_agreement("Arbel", "Elisheva", keep_singletons,True, file_by_annotator, conllu_annotated_files_path,)
Elisheva_vs_Ido =  two_annotators_agreement("Elisheva", "Ido", keep_singletons,True,file_by_annotator, conllu_annotated_files_path)

third_round_agreement = [
    ("Ido vs. Arbel No Singletons", Arbel_vs_Ido),
    ("Arbel vs. Elisheva  No Singletons", Arbel_vs_Elisheva),
    ("Elisheva vs. Ido No Singletons", Elisheva_vs_Ido)]
```

```python
third_round_agreement
```

```python
for name, score in third_round_agreement:
    print(name + " Conll score by muc, bcub, ceafe, lea  (without Mentions)")
    df = pd.DataFrame(score)
    df = df.drop(columns=["mentions"])
    df['mean'] = df.mean(axis=1)
    print(df['mean']['f1'])
for name, score in third_round_agreement:
    print(name + " Mention score")
    mentions = pd.DataFrame(score)["mentions"]
    print(mentions['f1'])
```

```python
should_keep_singleton = False
include_mentions = True
scores = n_annotators_agreement(["Arbel", "Ido", "Elisheva"], should_keep_singleton,include_mentions, file_by_annotator, conllu_annotated_files_path)
```

```python
print('Total score of "Arbel", "Ido", "Elisheva" with mentions: ')
data = {'values': [pd.DataFrame(score).loc['f1'].mean() for score in scores]}
df = pd.DataFrame(data)


print(df['values'].mean())
```

```python
should_keep_singleton = False
include_mentions = False
scores_no_mentions = n_annotators_agreement(["Arbel", "Ido", "Elisheva"], should_keep_singleton,include_mentions, file_by_annotator, conllu_annotated_files_path)
```

```python
print('Total score of "Arbel", "Ido", "Elisheva" no mentions: ')
data_no_mentions = {'values': [pd.DataFrame(score).loc['f1'].mean() for score in scores_no_mentions]}
df_no_mentions = pd.DataFrame(data_no_mentions)
print(df_no_mentions['values'].mean())
```
