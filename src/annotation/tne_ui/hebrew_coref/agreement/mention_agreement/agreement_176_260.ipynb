```python
file_by_annotator = read_annotation_data(conllu_annotated_files_path, specific_files=doc_range,
                                         exclude={'Shaked', 'Consolidation', 'Elisheva'})
```

```python
process_annotations(file_by_annotator)
```

```python
annotator_keys = ["Itay", "Hadar", "Tal"]

import itertools

for annotator_1, annotator_2 in itertools.combinations(annotator_keys, 2):
    pair_score = process_annotations(
        {key: file_by_annotator[key] for key in {annotator_1, annotator_2} if key in file_by_annotator})
    print(f"Agreement {annotator_1}-{annotator_2}: {pair_score:.3f}")
```
