```python
conllu_annotated_files_path = "/Users/s0g0a87/studies/tne_ui/hebrew_coref/Interview/coref_answers/training"
```

```python
files_id = set(list(range(71, 76)))
print(files_id)
file_by_annotator = defaultdict(list)
conllu_annotated_files = os.listdir(conllu_annotated_files_path)
for f in conllu_annotated_files:
    if int(f.split("_")[0]) in files_id:
        file_by_annotator[extract_name(f)].append(f)
keep_singletons = False
calc_with_mentions = False
#
# file_by_annotator = read_annotation_data(conllu_annotated_files_path, specific_files=set(list(range(65, 70))),
#                                          exclude={'Shaked', 'Arbel', 'Ido', 'Yuval Sagi', 'Bar Shwarts', 'Aviv Sason'})
```

```python
file_by_annotator
```

```python
agreement = {}
for ann in ['Hadar', 'Ariela', 'Itay']:
    for gold in ['Elisheva']:
        keys = {key: file_by_annotator[key] for key in [ann, gold] if key in file_by_annotator}
        print(f"{ann}:{gold}")
        agreement[f"{ann}:{gold}"] = two_annotators_agreement(ann, gold, keep_singletons, calc_with_mentions,
                                                              file_by_annotator, conllu_annotated_files_path, )
```
