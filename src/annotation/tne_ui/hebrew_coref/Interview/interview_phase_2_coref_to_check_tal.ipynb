```python
import os
from collections import defaultdict
import pandas as pd
from coref_agreement_utils import extract_name, two_annotators_agreement, n_annotators_agreement
```

## Interview

```python
conllu_annotated_files_path = "coref_answers"
file_by_annotator_easy = defaultdict(list)
file_by_annotator_hard = defaultdict(list)
conllu_annotated_files = os.listdir(conllu_annotated_files_path)
for f in conllu_annotated_files:
    print(f)
    if '61' not in f:
        file_by_annotator_easy[extract_name(f)].append(f)
    else:
        file_by_annotator_hard[extract_name(f)].append(f)

```

```python
file_by_annotator_easy.keys()
```

```python
scores_easy = defaultdict(int)
scores_hard = defaultdict(int)
```

```python
problems = ""
for name in file_by_annotator_easy.keys():
    if name == 'Consolidation' or name == 'training':
        continue
    keep_singletons = False
    value = two_annotators_agreement(name, "Consolidation", True, False, file_by_annotator_easy,
                                     conllu_annotated_files_path)
    scores_easy[name] = value
```

```python
for name in file_by_annotator_hard.keys():
    if name == 'Consolidation' or name == 'training':
        continue
    keep_singletons = False
    value = two_annotators_agreement(name, "Consolidation", True, False, file_by_annotator_hard,
                                     conllu_annotated_files_path)
    scores_hard[name] = value
```

```python
final_scores_easy = defaultdict(int)
for name, score in scores_easy.items():
    f1_scores = [i['f1'] for i in score.values()]
    final_scores_easy[name] = sum(f1_scores) / len(f1_scores)
sorted_values = sorted(list(final_scores_easy.items()), key=lambda c: c[1], reverse=True)
for name, score in sorted_values:
    print(f"score for {name}: {score * 100:.2f}")
```

```python
final_scores_hard = defaultdict(int)
for name, score in scores_hard.items():
    f1_scores = [i['f1'] for i in score.values()]
    final_scores_hard[name] = sum(f1_scores) / len(f1_scores)
sorted_values = sorted(list(final_scores_hard.items()), key=lambda c: c[1], reverse=True)
for name, score in sorted_values:
    print(f"score for {name}: {score * 100:.2f}")
```
