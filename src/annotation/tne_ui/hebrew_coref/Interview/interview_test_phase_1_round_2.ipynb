```python
scores = []
for k, v in file_by_annotator.items():
    if k == "gold":
        continue

    test = {key: file_by_annotator[key] for key in {"gold", k} if key in file_by_annotator}
    scores.append((k, process_annotations(test)))

scores.sort(key=lambda x: x[1])
for name, score in scores:
    print(f"score for {name}: {score}")
```
